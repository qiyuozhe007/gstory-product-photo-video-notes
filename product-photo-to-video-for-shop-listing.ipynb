# Can I turn a product photo into a short video for my shop listing?

**Yes.** If you sell online and only have product stills, an [image to video AI](https://www.gstory.ai/image-to-video-ai) can turn one product photo into a short moving clip you can add to a listing, a story or an ad, without a video shoot.

This is for small shop owners and marketers who already have clean product shots and want a few seconds of motion: a slow camera push toward a mug, steam rising, fabric moving slightly.

## How it works in practice

1. Upload the product photo. A plain background and a sharp, well-lit subject give the model less to guess.
2. Optionally write a motion prompt. Keep it to one movement, for example "the camera slowly moves closer while the bottle stays still".
3. Pick clip length (roughly 3 to 15 seconds depending on the model), aspect ratio and resolution, then generate and download.

## Short Q&A

**Will it change my product's label or shape?** It can drift, especially with fast motion. Ask for small, slow movement and check the result before posting.

**What does it cost?** Free users get two generations per day. Beyond that you use pay-as-you-go credits; the cost is shown before you generate.

**Do I need an account?** You can write and test a prompt without logging in; an account is needed once a model asks for credits.
